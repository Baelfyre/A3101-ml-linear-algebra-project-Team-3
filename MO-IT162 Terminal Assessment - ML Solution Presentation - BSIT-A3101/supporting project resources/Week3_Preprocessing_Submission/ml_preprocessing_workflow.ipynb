## Preprocessing Principles

The following principles will guide the preprocessing process:

1. The original FinMark datasets will remain unchanged.
2. Data-quality issues will be identified and documented before cleaning.
3. Separate cleaned DataFrames will be created for preprocessing.
4. Records containing missing values will be omitted rather than filled
   with unsupported estimated values.
5. Valid information will be standardized when differences are caused only
   by formatting, capitalization, spacing, or naming conventions.
6. Duplicate records will only be removed after they are verified.
7. Invalid or questionable values will be reviewed before deciding whether
   a record should be removed.
8. CustomerID, TransactionID, and InteractionID will be retained in the
   cleaned master datasets because they support record identification,
   relationships, duplicate checking, and data integrity.
9. Identifier fields will not be treated as machine learning features or
   ordinary analytical variables.
10. The effect of preprocessing will be measured by comparing the raw and
    cleaned datasets.

#### Date Format Pattern Review

The date strings are reviewed by general pattern so that different
representations can be identified before parsing.

This helps distinguish formatting inconsistency from invalid date values.

```python
import re

def detect_date_pattern(value):
    if pd.isna(value):
        return "Missing"

    value = str(value).strip()

    if re.fullmatch(r"\d{4}-\d{2}-\d{2}", value):
        return "YYYY-MM-DD"

    if re.fullmatch(r"\d{2}/\d{2}/\d{4}", value):
        return "DD/MM/YYYY or MM/DD/YYYY"

    return "Other"


signup_patterns = demographics_raw["SignupDate"].apply(detect_date_pattern)
transaction_patterns = transactions_raw["TransactionDate"].apply(detect_date_pattern)
interaction_patterns = social_raw["InteractionDate"].apply(detect_date_pattern)

print("SignupDate Patterns")
display(signup_patterns.value_counts(dropna=False))

print("\nTransactionDate Patterns")
display(transaction_patterns.value_counts(dropna=False))

print("\nInteractionDate Patterns")
display(interaction_patterns.value_counts(dropna=False))
```

#### Explicit Mixed-Date Parsing Test

The date fields may contain multiple valid representations.

To avoid interpreting valid formatting differences as invalid dates, a
temporary parsing function is used to test recognized date formats
separately.

The raw datasets remain unchanged.

```python
def parse_finmark_date(series):
    # Work with a temporary string copy
    date_strings = series.astype("string").str.strip()

    # Create an empty datetime Series
    parsed_dates = pd.Series(
        pd.NaT,
        index=series.index,
        dtype="datetime64[ns]"
    )

    # Format 1: YYYY-MM-DD
    iso_mask = date_strings.str.match(
        r"^\d{4}-\d{2}-\d{2}$",
        na=False
    )

    parsed_dates.loc[iso_mask] = pd.to_datetime(
        date_strings.loc[iso_mask],
        format="%Y-%m-%d",
        errors="coerce"
    )

    # Format 2: DD/MM/YYYY
    day_first_mask = date_strings.str.match(
        r"^\d{2}/\d{2}/\d{4}$",
        na=False
    )

    parsed_dates.loc[day_first_mask] = pd.to_datetime(
        date_strings.loc[day_first_mask],
        format="%d/%m/%Y",
        errors="coerce"
    )

    return parsed_dates
```

```python
signup_date_verified = parse_finmark_date(
    demographics_raw["SignupDate"]
)

transaction_date_verified = parse_finmark_date(
    transactions_raw["TransactionDate"]
)

interaction_date_verified = parse_finmark_date(
    social_raw["InteractionDate"]
)
```

### Conflicting Identifier Review

Exact duplicate rows have already been identified. However, record-level
identifiers that are expected to be unique must also be checked after exact
duplicates are removed.

The following identifiers are expected to represent one record:

- `CustomerID` in Customer Demographics
- `TransactionID` in Customer Transactions
- `InteractionID` in Social Media Interactions

If the same identifier appears in multiple non-identical records, the
records are treated as conflicting because the available data does not
provide sufficient evidence to determine which version is authoritative.

No conflicting record is automatically retained based only on row order.

```python
def find_conflicting_identifiers(df, id_column):
    deduplicated = df.drop_duplicates()

    conflict_mask = deduplicated.duplicated(
        subset=id_column,
        keep=False
    )

    return (
        deduplicated.loc[conflict_mask]
        .sort_values(id_column)
    )


demographic_conflicts = find_conflicting_identifiers(
    demographics_raw,
    "CustomerID"
)

transaction_conflicts = find_conflicting_identifiers(
    transactions_raw,
    "TransactionID"
)

social_conflicts = find_conflicting_identifiers(
    social_raw,
    "InteractionID"
)


conflict_summary = pd.DataFrame({
    "Dataset": [
        "Customer Demographics",
        "Customer Transactions",
        "Social Media Interactions"
    ],
    "Conflicting Identifiers": [
        demographic_conflicts["CustomerID"].nunique(),
        transaction_conflicts["TransactionID"].nunique(),
        social_conflicts["InteractionID"].nunique()
    ],
    "Affected Rows": [
        len(demographic_conflicts),
        len(transaction_conflicts),
        len(social_conflicts)
    ]
})

display(conflict_summary)
```

This avoids arbitrarily retaining one conflicting record based only on

## Data Cleaning

Separate working copies are created from the verified raw datasets.

All cleaning operations are applied only to these working copies. The
original raw DataFrames remain unchanged so that the preprocessing process
can be audited and reproduced.

```python
# CREATE CLEANED WORKING COPIES
# Separate copies of the raw datasets are created before cleaning.
# All changes will be applied only to the _clean DataFrames so the
# original raw datasets remain unchanged for comparison and auditing.

demographics_clean = demographics_raw.copy()
transactions_clean = transactions_raw.copy()
social_clean = social_raw.copy()

print("Cleaned working copies created.")
```

```python
# CREATE CLEANING AUDIT LOG
# This log records the number of rows before and after each cleaning step.
# It provides evidence of how preprocessing affects each dataset and will
# support the final before-versus-after summary.

cleaning_log = []

def log_cleaning_step(dataset, step, before_rows, after_rows):
    cleaning_log.append({
        "Dataset": dataset,
        "Cleaning Step": step,
        "Rows Before": before_rows,
        "Rows After": after_rows,
        "Rows Removed": before_rows - after_rows
    })

print("Cleaning audit log initialized.")
```

```python
# NORMALIZE COLUMN NAMES
# Column names are converted to snake_case for consistent and readable
# Pandas operations. Only column labels are changed. Dataset values remain
# unchanged.

import re

def to_snake_case(column_name):
    first_pass = re.sub(
        r"(.)([A-Z][a-z]+)",
        r"\1_\2",
        column_name.strip()
    )

    return re.sub(
        r"([a-z0-9])([A-Z])",
        r"\1_\2",
        first_pass
    ).lower()


demographics_clean.columns = [
    to_snake_case(column)
    for column in demographics_clean.columns
]

transactions_clean.columns = [
    to_snake_case(column)
    for column in transactions_clean.columns
]

social_clean.columns = [
    to_snake_case(column)
    for column in social_clean.columns
]

print("Customer Demographics:")
print(demographics_clean.columns.tolist())

print("\nCustomer Transactions:")
print(transactions_clean.columns.tolist())

print("\nSocial Media Interactions:")
print(social_clean.columns.tolist())
```

```python
# REMOVE CONFIRMED EXACT DUPLICATES
# Fully identical rows are removed because they repeat the same information
# and may overrepresent observations during later analysis.
# The number of removed records is recorded in the cleaning audit.

before = len(demographics_clean)

demographics_clean = (
    demographics_clean
    .drop_duplicates()
    .copy()
)

log_cleaning_step(
    "Customer Demographics",
    "Remove exact duplicate rows",
    before,
    len(demographics_clean)
)


before = len(transactions_clean)

transactions_clean = (
    transactions_clean
    .drop_duplicates()
    .copy()
)

log_cleaning_step(
    "Customer Transactions",
    "Remove exact duplicate rows",
    before,
    len(transactions_clean)
)


before = len(social_clean)

social_clean = (
    social_clean
    .drop_duplicates()
    .copy()
)

log_cleaning_step(
    "Social Media Interactions",
    "Remove exact duplicate rows",
    before,
    len(social_clean)
)

print("Exact duplicate removal completed.")
```

```python
# REMOVE UNRESOLVED IDENTIFIER CONFLICTS
# Record-level identifiers are expected to represent one unique record.
# When the same identifier belongs to multiple non-identical records and
# there is no evidence showing which record is authoritative, the entire
# conflicting identifier group is omitted rather than arbitrarily keeping
# the first occurrence.

def remove_conflicting_ids(df, id_column):
    conflicting_ids = df.loc[
        df.duplicated(
            subset=id_column,
            keep=False
        ),
        id_column
    ].unique()

    cleaned = df[
        ~df[id_column].isin(conflicting_ids)
    ].copy()

    return cleaned, conflicting_ids


before = len(demographics_clean)

demographics_clean, demographic_conflict_ids = (
    remove_conflicting_ids(
        demographics_clean,
        "customer_id"
    )
)

log_cleaning_step(
    "Customer Demographics",
    "Remove unresolved CustomerID conflicts",
    before,
    len(demographics_clean)
)


before = len(transactions_clean)

transactions_clean, transaction_conflict_ids = (
    remove_conflicting_ids(
        transactions_clean,
        "transaction_id"
    )
)

log_cleaning_step(
    "Customer Transactions",
    "Remove unresolved TransactionID conflicts",
    before,
    len(transactions_clean)
)


before = len(social_clean)

social_clean, social_conflict_ids = (
    remove_conflicting_ids(
        social_clean,
        "interaction_id"
    )
)

log_cleaning_step(
    "Social Media Interactions",
    "Remove unresolved InteractionID conflicts",
    before,
    len(social_clean)
)

print("Identifier conflict treatment completed.")
```

```python
# OMIT RECORDS CONTAINING MISSING VALUES
# Missing values are not imputed because the supplied FinMark datasets do
# not provide sufficient evidence for reliable replacement values.
# Rows containing missing information are therefore omitted, and the
# resulting reduction in sample size is documented.

before = len(demographics_clean)

demographics_clean = (
    demographics_clean
    .dropna()
    .copy()
)

log_cleaning_step(
    "Customer Demographics",
    "Omit rows containing missing values",
    before,
    len(demographics_clean)
)


before = len(transactions_clean)

transactions_clean = (
    transactions_clean
    .dropna()
    .copy()
)

log_cleaning_step(
    "Customer Transactions",
    "Omit rows containing missing values",
    before,
    len(transactions_clean)
)


before = len(social_clean)

social_clean = (
    social_clean
    .dropna()
    .copy()
)

log_cleaning_step(
    "Social Media Interactions",
    "Omit rows containing missing values",
    before,
    len(social_clean)
)

print("Missing-value treatment completed.")
```

```python
# CLEAN CUSTOMER AGE
# Age is converted to numeric form.
# Records are retained only when Age can be converted successfully and
# falls within the documented analytical range of 0 to 120.
# Values such as Unknown, negative ages, and ages above 120 are omitted
# rather than estimated or corrected.

before = len(demographics_clean)

demographics_clean["age"] = pd.to_numeric(
    demographics_clean["age"],
    errors="coerce"
)

valid_age_mask = (
    demographics_clean["age"].notna()
    & demographics_clean["age"].between(
        0,
        120,
        inclusive="both"
    )
)

demographics_clean = (
    demographics_clean.loc[
        valid_age_mask
    ]
    .copy()
)

demographics_clean["age"] = (
    demographics_clean["age"]
    .astype("int64")
)

log_cleaning_step(
    "Customer Demographics",
    "Remove unusable or invalid Age records",
    before,
    len(demographics_clean)
)

print("Age cleaning completed.")
```

```python
# CLEAN TRANSACTION AMOUNT
# Amount is converted to numeric form.
# "Free" is normalized to 0.0 because it explicitly represents no monetary
# charge rather than an unknown amount.
# Negative amounts are omitted because no transaction-type field supports
# interpreting them as refunds or reversals.
# Existing numeric zero values are retained.

before = len(transactions_clean)

amount_text = (
    transactions_clean["amount"]
    .astype("string")
    .str.strip()
)

free_mask = (
    amount_text
    .str.casefold()
    .eq("free")
)

amount_text.loc[free_mask] = "0"

transactions_clean["amount"] = pd.to_numeric(
    amount_text,
    errors="coerce"
)

valid_amount_mask = (
    transactions_clean["amount"].notna()
    & (transactions_clean["amount"] >= 0)
)

transactions_clean = (
    transactions_clean.loc[
        valid_amount_mask
    ]
    .copy()
)

transactions_clean["amount"] = (
    transactions_clean["amount"]
    .astype("float64")
)

log_cleaning_step(
    "Customer Transactions",
    "Normalize Free to 0 and remove unsupported negative Amount records",
    before,
    len(transactions_clean)
)

print("Transaction Amount cleaning completed.")
```

```python
# NORMALIZE DATE FIELDS
# The verified mixed date formats are converted using the explicit FinMark
# date parser developed during verification.
# All valid dates are stored as proper datetime values and will later be
# exported consistently in YYYY-MM-DD format.

demographics_clean["signup_date"] = parse_finmark_date(
    demographics_clean["signup_date"]
)

transactions_clean["transaction_date"] = parse_finmark_date(
    transactions_clean["transaction_date"]
)

social_clean["interaction_date"] = parse_finmark_date(
    social_clean["interaction_date"]
)

print(
    "Unparsed SignupDate values:",
    demographics_clean["signup_date"].isna().sum()
)

print(
    "Unparsed TransactionDate values:",
    transactions_clean["transaction_date"].isna().sum()
)

print(
    "Unparsed InteractionDate values:",
    social_clean["interaction_date"].isna().sum()
)
```

```python
# RESET CLEANED DATASET INDEXES
# Row removal leaves gaps in the original Pandas indexes.
# Resetting creates a continuous zero-based index without modifying the
# actual CustomerID, TransactionID, or InteractionID fields.

demographics_clean = (
    demographics_clean
    .reset_index(drop=True)
)

transactions_clean = (
    transactions_clean
    .reset_index(drop=True)
)

social_clean = (
    social_clean
    .reset_index(drop=True)
)

print("Cleaned dataset indexes reset.")
```

```python
# RELOAD SAVED OUTPUT FILES
# The exported CSV files are loaded again to verify that the actual files
# intended for submission remain complete and usable after serialization.

demographics_final = pd.read_csv(
    demographics_output
)

transactions_final = pd.read_csv(
    transactions_output
)

social_final = pd.read_csv(
    social_output
)

print("Saved cleaned datasets reloaded successfully.")
```

```python
# VALIDATE EXPORTED WEEK 3 FILES
# The reloaded CSV files are checked for row-count consistency, missing
# values, and exact duplicate records.
#
# This ensures that the submitted files match the validated in-memory
# cleaned datasets.

saved_output_validation = pd.DataFrame({
    "Dataset": [
        "Customer Demographics",
        "Customer Transactions",
        "Social Media Interactions"
    ],

    "Expected Rows": [
        len(demographics_clean),
        len(transactions_clean),
        len(social_clean)
    ],

    "Reloaded Rows": [
        len(demographics_final),
        len(transactions_final),
        len(social_final)
    ],

    "Missing Values": [
        demographics_final.isna().sum().sum(),
        transactions_final.isna().sum().sum(),
        social_final.isna().sum().sum()
    ],

    "Exact Duplicates": [
        demographics_final.duplicated().sum(),
        transactions_final.duplicated().sum(),
        social_final.duplicated().sum()
    ]
})

saved_output_validation["Row Count Match"] = (
    saved_output_validation["Expected Rows"]
    ==
    saved_output_validation["Reloaded Rows"]
)

display(saved_output_validation)
```

```python
# VERIFY EXPORTED DATE FORMAT
# Date fields in the saved CSV files are checked to confirm that every
# value follows the finalized YYYY-MM-DD format.

def valid_iso_dates(series):
    return (
        series
        .astype("string")
        .str.fullmatch(
            r"\d{4}-\d{2}-\d{2}"
        )
        .all()
    )


date_export_validation = pd.DataFrame({
    "Date Field": [
        "signup_date",
        "transaction_date",
        "interaction_date"
    ],

    "YYYY-MM-DD Format Valid": [
        valid_iso_dates(
            demographics_final["signup_date"]
        ),
        valid_iso_dates(
            transactions_final["transaction_date"]
        ),
        valid_iso_dates(
            social_final["interaction_date"]
        )
    ]
})

display(date_export_validation)
```

```python
# FINAL WEEK 3 OUTPUT VALIDATION
# A single final result confirms whether the exported files satisfy the
# core preprocessing output checks required before submission.

week3_outputs_valid = (
    saved_output_validation["Row Count Match"].all()
    and (saved_output_validation["Missing Values"] == 0).all()
    and (saved_output_validation["Exact Duplicates"] == 0).all()
    and date_export_validation[
        "YYYY-MM-DD Format Valid"
    ].all()
)

print(
    "Week 3 finalized output validation passed:",
    week3_outputs_valid
)
```
